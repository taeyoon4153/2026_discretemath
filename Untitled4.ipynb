```python
#예시 그래프: 친구관계
graph={
    'Alice':['Bob','Charlie'],
    'Bob':['David','Eve'],
    'Charlie':['Frank'],
    'David':[],
    'Eve':['Frank'],
    'Frank':[]
}

```

```python
#수도코드 이건 할당하지말기
void dfs (G,v)
{
    visited[v]=TRUE;
    print(v is visited);
    for each vertex w adjacent to v
      if (visited[w]==FALSE)
        dfs(G,w);
}

void DFS(G) //G=(V,E)
{
    for all u in V
      visited[u]=FALSE;
    for each vertex u in V
      if (visited[u]==FALSE)
        dfs(G,u);
}
```

```python
def dfs(graph,v,visited):
  visited.add(v)
  print(v,end=' ')
  for neighbor in visited:
    if neighbor not in visited:
      dfs(graph,neighbor,visited)
```

```python
print("DFS 탐색 순서:")
dfs(graph,'Alice',set())
```

```python
#큐(Queue)를 효율적으로 구현할 수 있는 도구
from collections import deque

def bfs(graph,start):
  visited=set()
  queue=deque([start])
  visited.add(start)

  while queue:
    v=queue.popleft()
    print(v, end='')
    for neighbor in graph[v]:
      if neighbor not in visited:
        visited.add(neighbor)
        queue.append(neighbor)
print("\nBFS 탐색 순서:")
bfs(graph,'Alice')
```

```python
from collections import deque

def recommend_friend(graph,start):
  visited=set()
  queue=deque([(start, 0)])   #(노드,거리)

  print(f"\n{start}의 친구 추천 (2단계):")

  while queue:
    current, depth=queue.popleft()
    if current not in visited:
      visited.add(current)
      if depth ==2:
        print("-",current)
      if depth<2:
        for neighbor in graph[current]:
          queue.append((neighbor,depth+1))
```

```python
recommend_friend(graph,'Alice')
```
