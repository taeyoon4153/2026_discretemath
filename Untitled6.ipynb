```python
graph={
    '김태윤':['루비','김예린','김유진','이세영','마은영','조은비'],
    '이세영':['임하린', '김나영', '김민서','김태윤','마은영'],
    '마은영':['이태림', '이예원', '주민지','김태윤','이세영','오선우','조은비','김유미','권휘'],
    '오선우':['조은비', '유재윤','김유미','마은영','권휘'],
    '권휘':['구정남', '강민서', '유병준','마은영','오선우'],
}
```

```python
from collections import deque

def bfs(graph,start):
  visited=set()
  queue=deque([start])
  visited.add(start)

  while queue:
    v=queue.popleft()
    print(v, end=' ')
    for neighbor in graph.get(v, []):
      if neighbor not in visited:
        visited.add(neighbor)
        queue.append(neighbor)
print("\nBFS 탐색 순서:")
bfs(graph,'김태윤')
```

```python
graph={
    '김태윤':['루비','김예린','김유진','이세영','마은영','조은비'],
    '이세영':['임하린', '김나영', '김민서','김태윤','마은영'],
    '마은영':['이태림', '이예원', '주민지','김태윤','이세영','오선우','조은비','김유미','권휘'],
    '오선우':['조은비', '유재윤','김유미','마은영','권휘'],
    '권휘':['구정남', '강민서', '유병준','마은영','오선우'],
}
```

```python
from collections import deque

def contagion_simulation(graph,start):
  visited=set()
  queue=deque([(start, 0)])   #(노드,거리)

  print(f"\n{start}과의 접촉자:")

  while queue:
    current, depth=queue.popleft()
    if current not in visited:
      visited.add(current)
      if 0<depth<=2:
        print("-",current,"감염자 경로의 거리",depth)
      if depth<2:
        for neighbor in graph.get(current, []):
          queue.append((neighbor,depth+1))
```

```python
contagion_simulation(graph,"마은영")
```

```python
def contagion_simulation(graph,start):
  visited=set()
  queue=deque([(start,0)])

  print(f"\n감염자 {start} 접촉자")

  while queue:
    current,depth=queue.popleft()
    if current not in visited:
      visited.add(current)
      if 0<depth<=2:
        print("-",current,"감염경로거리",depth)
      if depth<2:
        for i in graph.get(current, []):
          queue.append((i,depth+1))

contagion_simulation(graph,"김태윤")
```
